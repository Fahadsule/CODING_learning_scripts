Creating an array but without unneccesary copying.

```python
import numpy as np
data=np.array([0.67,0.89,0.78])
data2=np.asarray(data)
data2[0]=0.798
print(data)# As you can see the change applied to data2 directly affects data ,so it was never a copy just a reassigment. 
```

Creating an array of zeros


```python
import numpy as np
data=np.zeros((10,5))
print(data)
```

CREATING AN ARRAY OF ZEROS THAT REPLICATES THE SAME DIMENSIONS AND SIZE AS A PREVIOUS

```python
import numpy as np
data=np.array([[1,3, 4.5],[2,7,8]])
zero=np.zeros_like(data)
print(zero)
```

Building an array with values in a range

```python
import numpy as np
data=np.arange(3,15,3)# Pretty much works the same as the normal python one
print(data)
```

Making a list of ones

```python
import numpy as np
data=np.ones((3,7))
print(data)
```

Making an array of ones but replicating the structure of the other array


```python
import numpy as np
data=np.array([[1,3, 4.5],[2,7,8]])
one=np.ones_like(data)
print(one)
```

To create an array and not populte with any values

```python
import numpy as np
data4=np.empty((2,4))#Simply allocates the memory
print(data4)# Returns what was stored previously on that location so probably some random values
```

To simply allocate memory space and replicate the shpe and size of the other array

You can convert from one data type to the other


```python
import numpy as np
data=np.array([1,3,4,5])#This is type int64
data.dtype
#To convert thy data type the .astype() method
data2=data.astype(np.float64)
data2.dtype
```

We can also use another array's dtype attribute and replicate that directly

```python
import numpy as np
data=np.array([2,3,4,5,5], dtype=np.float64)
int_array=np.array([2,2,3,3,4,7])
int_array=int_array.astype(data.dtype)
int_array.dtype
```
